# Async in Python — A Simple Demo

This notebook walks through how `async`/`await` works in Python using `asyncio`.

**Key concepts:**
- `async def` — defines a coroutine function
- `await` — suspends the coroutine until the awaited task is done, yielding control back to the event loop
- `asyncio.gather()` — runs multiple coroutines concurrently

The central idea: while one coroutine is *waiting* (e.g., for I/O or a simulated delay), the event loop runs other coroutines — all in a **single thread**.

## 1. Synchronous baseline — tasks run one after another

````python
import time

def fetch_data_sync(name: str, delay: float) -> str:
    """Simulates a blocking I/O call (e.g. HTTP request, DB query)."""
    print(f"[sync]  START  {name}")
    time.sleep(delay)          # blocks the whole program
    print(f"[sync]  DONE   {name} (after {delay}s)")
    return f"{name} result"

start = time.perf_counter()

fetch_data_sync("Task-A", 2)
fetch_data_sync("Task-B", 1)
fetch_data_sync("Task-C", 1)

elapsed = time.perf_counter() - start
print(f"\nTotal time (sync): {elapsed:.2f}s  <- sum of all delays")
````

## 2. Async version — tasks run concurrently

````python
import asyncio

async def fetch_data_async(name: str, delay: float) -> str:
    """Simulates a non-blocking I/O call."""
    print(f"[async] START  {name}")
    await asyncio.sleep(delay)   # yields control; other coroutines can run
    print(f"[async] DONE   {name} (after {delay}s)")
    return f"{name} result"

start = time.perf_counter()

# asyncio.gather runs all three coroutines concurrently
results = await asyncio.gather(
    fetch_data_async("Task-A", 2),
    fetch_data_async("Task-B", 1),
    fetch_data_async("Task-C", 1),
)

elapsed = time.perf_counter() - start
print(f"\nResults : {results}")
print(f"Total time (async): {elapsed:.2f}s  <- dominated by the longest task only")
````

## 3. What's actually happening?

```
Timeline (async)
----------------------------------------------------
t=0.0s  Task-A starts  (delay=2s)
t=0.0s  Task-B starts  (delay=1s)  <- event loop switches immediately
t=0.0s  Task-C starts  (delay=1s)  <- event loop switches again
t=1.0s  Task-B done
t=1.0s  Task-C done
t=2.0s  Task-A done
----------------------------------------------------
Total: ~2s  (not 4s like the sync version)
```

**Rule of thumb:** `async`/`await` shines for **I/O-bound** work (network, disk, DB).  
For **CPU-bound** work use `multiprocessing` or `concurrent.futures.ProcessPoolExecutor` instead.

## 4. Bonus — `asyncio.create_task` for fire-and-forget style

````python
async def pipeline() -> None:
    """Schedule tasks independently and await them later."""
    # Schedule tasks without awaiting immediately
    task_a = asyncio.create_task(fetch_data_async("Task-A", 2), name="A")
    task_b = asyncio.create_task(fetch_data_async("Task-B", 1), name="B")

    print("Tasks are scheduled - doing other work in the meantime...")
    await asyncio.sleep(0)   # yield once so tasks can start

    # Now wait for both to finish
    result_a = await task_a
    result_b = await task_b
    print(f"\nFinal results: {result_a}, {result_b}")

await pipeline()
````
